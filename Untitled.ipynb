```python
import torch
from transformers import Qwen2VLForConditionalGeneration, AutoProcessor
from PIL import Image
import matplotlib.pyplot as plt
import numpy as np
import requests

# Load model and processor
model = Qwen2VLForConditionalGeneration.from_pretrained(
    "Qwen/Qwen2-VL-7B-Instruct", torch_dtype=torch.float16, device_map="auto"
)
processor = AutoProcessor.from_pretrained("Qwen/Qwen2-VL-7B-Instruct")

# Store attention maps
attention_maps = []

# Register hooks to capture attention
def get_attention_hook(layer_idx):
    def hook(module, input, output):
        # The attention output contains the attention weights in Transformer models
        # Exact structure may vary by implementation, but usually it's in output[1]
        if isinstance(output, tuple) and len(output) > 1:
            attn_map = output[1]  # This is typically where attention weights are
            attention_maps.append((layer_idx, attn_map.detach().cpu()))
    return hook

# Process image and text
url = "https://qianwen-res.oss-cn-beijing.aliyuncs.com/Qwen-VL/assets/demo.jpeg"
image = Image.open(requests.get(url, stream=True).raw)

conversation = [
    {"role": "user", "content": [{"type": "image"}, {"type": "text", "text": "Describe this image."}]}
]
text_prompt = processor.apply_chat_template(conversation, add_generation_prompt=True)
inputs = processor(text=[text_prompt], images=[image], padding=True, return_tensors="pt")
inputs = {k: v.to(model.device) for k, v in inputs.items()}

# First, run a forward pass to analyze token structure
with torch.no_grad():
    # Enable output_attentions to see if the model provides attention natively
    outputs = model(**inputs, output_attentions=True)

# Check if the model provides attention maps directly
if hasattr(outputs, 'attentions') and outputs.attentions:
    print("Model provides attention maps directly!")
    # Use the model's native attention outputs
    attention_maps = [(i, attn) for i, attn in enumerate(outputs.attentions)]
else:
    print("Adding hooks to capture attention...")
    # Register hooks for each layer
    hooks = []
    for i, block in enumerate(model.model.layers):
        # This path might need adjustment for Qwen2-VL
        if hasattr(block, 'self_attn'):
            hook = block.self_attn.register_forward_hook(get_attention_hook(i))
            hooks.append(hook)
        
    # Run forward pass again with hooks
    with torch.no_grad():
        outputs = model(**inputs)
        
    # Remove hooks
    for hook in hooks:
        hook.remove()
```

```python
# Analyze the attention maps properly
if attention_maps:
    # Get the dimensions of the first attention map to understand the structure
    example_map = attention_maps[0][1]
    print(f"Attention map shape: {example_map.shape}")
    
    # Calculate potential image token count from input shape
    input_shape = inputs['pixel_values'].shape
    print(f"Input pixel values shape: {input_shape}")
    
    # Determine text token count
    text_token_count = inputs['input_ids'].shape[1]
    print(f"Text token count: {text_token_count}")
    
    # Visualize attention maps for a few layers
    for layer_idx, attn_map in attention_maps[:5]:  # Just show first 5 layers
        # The attention map shape is typically [batch, heads, seq_len, seq_len]
        # We can average across attention heads
        avg_attention = attn_map.mean(dim=1)[0]  # Average heads, take first batch
        
        plt.figure(figsize=(12, 10))
        plt.imshow(avg_attention.cpu().numpy(), cmap='viridis')
        plt.colorbar()
        plt.title(f"Layer {layer_idx} Average Attention")
        
        # Add markers for text/image boundary if we know it
        if text_token_count < avg_attention.shape[0]:
            plt.axhline(y=text_token_count, color='r', linestyle='-', alpha=0.5)
            plt.axvline(x=text_token_count, color='r', linestyle='-', alpha=0.5)
            
        plt.savefig(f"layer_{layer_idx}_attention.png")
        plt.close()
        
        # Extract text-to-image attention if possible
        if text_token_count < avg_attention.shape[0]:
            text_to_img = avg_attention[:text_token_count, text_token_count:]
            img_to_text = avg_attention[text_token_count:, :text_token_count]
            
            # Visualize text-to-image attention
            plt.figure(figsize=(12, 6))
            plt.subplot(1, 2, 1)
            plt.imshow(text_to_img.numpy(), cmap='viridis')
            plt.title(f"Layer {layer_idx} Text-to-Image Attention")
            plt.colorbar()
            
            # Visualize image-to-text attention
            plt.subplot(1, 2, 2)
            plt.imshow(img_to_text.numpy(), cmap='viridis')
            plt.title(f"Layer {layer_idx} Image-to-Text Attention")
            plt.colorbar()
            
            plt.tight_layout()
            plt.savefig(f"layer_{layer_idx}_cross_attention.png")
            plt.close()
else:
    print("No attention maps captured. Check model architecture or hook placement.")
```
